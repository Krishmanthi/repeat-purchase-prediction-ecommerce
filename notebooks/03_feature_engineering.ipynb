```python
import pandas as pd

df = pd.read_csv(
    "../data/processed/customer_orders.csv"
)

df.head()
```

```python
df["order_purchase_timestamp"] = pd.to_datetime(
    df["order_purchase_timestamp"]
)
```

```python
snapshot_date = (
    df["order_purchase_timestamp"].max()
    + pd.Timedelta(days=1)
)

snapshot_date
```

```python
rfm = df.groupby("customer_unique_id").agg({
    "order_purchase_timestamp": lambda x:
        (snapshot_date - x.max()).days,
    "order_id": "count",
    "payment_value": "sum"
})
```

```python
rfm.columns = [
    "Recency",
    "Frequency",
    "Monetary"
]

rfm.head()
```

```python
journey_features = df.groupby(
    "customer_unique_id"
).agg({
    "delivery_days":"mean",
    "delivery_delay":"mean",
    "review_score":"mean",
    "installments":"mean",
    "total_freight":"mean",
    "payment_value":"mean"
})
```

```python
journey_features.columns = [
    "avg_delivery_days",
    "avg_delivery_delay",
    "avg_review_score",
    "avg_installments",
    "avg_freight",
    "avg_order_value"
]
```

```python
purchase_count = (
    df.groupby("customer_unique_id")
      .size()
)
```

```python
repeat_purchase = (
    purchase_count > 1
).astype(int)
```

```python
repeat_purchase.value_counts()
```

```python
customer_df = rfm.join(
    journey_features
)

customer_df["repeat_purchase"] = repeat_purchase
```
